```python
input_folder = "/media/arafat/New Volume/UrbanSound8K/UrbanSound8K/"
output_folder = "/media/arafat/New Volume/UrbanSound8K_Organized"

class_mapping = {
    0: "air_conditioner",
    1: "car_horn",
    2: "children_playing",
    3: "dog_bark",
    4: "drilling",
    5: "engine_idling",
    6: "gun_shot",
    7: "jackhammer",
    8: "siren",
    9: "street_music"
}
```

```python
"""
This script organizes the UrbanSound8K dataset into a format that is easier to work with.
It creates a folder for each class in the dataset, and moves all files associated with a class.

The UrbanSound8K dataset is divided into 10 folders, one for each class. The naming convention
"""
import os
import shutil

for class_name in class_mapping.values():
    os.makedirs(os.path.join(output_folder, class_name), exist_ok=True)
    
    
# read the audio files
folders = os.listdir(input_folder+"/audio")
for folder in folders:
    # get the files in the folder
    if not folder.startswith("fold"):
        continue
    files = os.listdir(input_folder+"/audio/"+folder)
    # loop through the files
    for f in files:
        if not f.endswith(".wav"):
            continue
        class_number = int(f.split("-")[1])
        
        class_name = class_mapping.get(class_number)
        
        source_path = os.path.join(input_folder+"audio", folder, f)
        destination_path = os.path.join(output_folder, class_name, f)
        # print(source_path, destination_path)
        shutil.copy2(source_path, destination_path)
    

```
